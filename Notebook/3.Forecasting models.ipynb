```python
def prepare_data(df1,target="Electricity_Price_CAD_MWh"):

    df1=df1.sort_index()

    # missing
    df1=df1.ffill().bfill()

    y=df1[target]

    X=df1.select_dtypes(include="number").copy()

    leakage_cols=[
        "Energy_Demand_MW",
        "Energy_Price_CAD",
        "Electricity_Price_CAD_MWh",
        "Gas_Price_CAD_GJ"]

    X=X.drop(columns=leakage_cols,errors="ignore")

    # =========================
    # IMPORTANT FIX (NaN + Inf)
    # =========================

    X=X.replace([np.inf,-np.inf],np.nan)
    X=X.fillna(0)
    y=y.ffill().bfill()

    # split
    split=int(len(df1)*0.8)

    X_train=X.iloc[:split]
    X_test=X.iloc[split:]

    y_train=y.iloc[:split]
    y_test=y.iloc[split:]

    return X_train,X_test,y_train,y_test
```

```python
def evaluate(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2
```

  <p style="font-size:140%; color:#556b2f; font-family:Comic Sans MS; line-height:1.4;">

    • Best when data has <b>no strong seasonality</b> but clear trends.<br>

  <h2 style="color:blue;">2. SARIMA Model</h2>
  <p style="font-size:140%; color:#556b2f; font-family:Comic Sans MS; line-height:1.4;">
    • <b>SARIMA</b> extends ARIMA by capturing <b>seasonality</b> explicitly.<br>
    • Uses <b>(p, d, q)</b> and seasonal <b>(P, D, Q, m)</b> parameters.<br>
    • Suitable for weekly sales with annual seasonal cycles.<br>
    • Produces stable multi-step forecasts with confidence intervals.
  </p>
  <hr>

  <!-- --------------------- LSTM SECTION -------------------- -->
  <h2 style="color:blue;">3. LSTM (Long Short-Term Memory)</h2>
  <p style="font-size:140%; color:#556b2f; font-family:Comic Sans MS; line-height:1.4;">
    • An advanced <b>deep learning RNN</b> capable of capturing long-range dependencies.<br>
    • Learns patterns through memory cells and gates.<br>
    • Requires sliding-window supervised learning (lag features).<br>
    • Performs extremely well when data has <b>complex nonlinear relationships</b>.
  </p>
  <hr>

  <!-- --------------------- RNN SIMPLE SECTION -------------------- -->
  <h2 style="color:blue;">4. Vanilla RNN</h2>
  <p style="font-size:140%; color:#556b2f; font-family:Comic Sans MS; line-height:1.4;">
    • A simpler recurrent model compared to LSTM.<br>
    • Models sequential patterns but struggles with long-term memory.<br>
    • Good baseline to compare with LSTM/GRU.<br>
    • Often shows higher error due to vanishing gradient issues.
  </p>

  <h2>📈 Stationarity</h2>

  <p>
    <b>In time series analysis, stationarity refers to a property of a series where its statistical properties 
    (like mean, variance, and autocorrelation) are constant over time.</b>
  </p>

  <p>
    Stationarity plays a vital role in time series analysis. When we grasp this concept, we gain insights into 
    which methods are suitable for our data and how we can enhance forecast accuracy. By converting non-stationary 
    data, we can apply various statistical tools to understand and predict time series behavior.
  </p>

  <p>
    Imagine you're trying to predict the weather. Stationarity is like having consistent weather patterns. 
    If it's usually sunny in summer and rainy in winter, that's stationary. But if the weather goes crazy with 
    random heat waves and snowstorms, that's non-stationary.
  </p>

  <p>
    Many forecasting tools work best with consistent patterns. By smoothing out wild swings in your data 
    (making the “crazy weather” more predictable), you can make better forecasts for your time series.
  </p>

  <p style="text-align:center;">
    <img src="https://miro.medium.com/v2/resize:fit:1147/1*xdblkZyg6YmmReAkZHUksw.png" 
         alt="stationarity" style="max-width:100%; height:auto; border-radius:5px;">
  </p>

  <h3>🔹 Intuitive Explanation</h3>
  <p>
    Imagine a time series like a river flowing at a constant rate. The average water level (mean) stays the same, 
    and the fluctuations around that average (variance) are consistent throughout the river's course. 
    This is a stationary series.
  </p>
  <p>
    In contrast, a non-stationary series would be like a river with changing water levels — flooding seasonally 
    or drying up over time. The mean and variability change depending on the season or long-term trends.
  </p>

  <h3>🔹 Formal Definition</h3>
  <p>A time series is stationary if:</p>
  <ul>
    <li><b>Mean is constant:</b> The average value of the series does not trend upward or downward.</li>
    <li><b>Variance is constant:</b> The spread of the data points around the mean remains consistent.</li>
    <li><b>Autocorrelation is constant:</b> Relationships between observations at different lags do not change over time.</li>
  </ul>

  <h3>🔹 Why Stationarity Matters</h3>
  <p>
    Many statistical methods for time series forecasting assume stationarity. Non-stationary data may produce unreliable results:
  </p>
  <ul>
    <li><b>Forecasting:</b> Predicting future values is difficult if patterns change over time.</li>
    <li><b>Statistical Tests:</b> Validity of tests often relies on stationary data.</li>
  </ul>

  <h3>🔹 Transforming Non-Stationary Data</h3>
  <p>Common techniques to make a series stationary include:</p>
  <ul>
    <li><b>Differencing:</b> Subtracting a previous value from the current value to remove trends.</li>
    <li><b>Detrending:</b> Fitting a trend line and subtracting it from the series to remove long-term trends.</li>
    <li><b>Seasonal Adjustment:</b> Seasonal differencing or adjustments remove seasonal patterns.</li>
  </ul>

      📊 Augmented Dickey-Fuller (ADF) Test
  </p>

  <p>
      The <strong>Augmented Dickey-Fuller (ADF) test</strong> is a statistical test used to determine whether a time series is 
      <strong>stationary or non-stationary</strong>. Stationarity is extremely important in many time series models such as ARIMA.
  </p>

  <p>
     The ADF test evaluates the <strong>null hypothesis</strong> that the time series contains a <em>unit root</em>, meaning 
     that it is <strong>non-stationary</strong>. The alternative hypothesis suggests the series <strong>is stationary</strong>.
  </p>

  <p>
     When running the ADF test, two values are mainly interpreted:
  </p>

  <ul style="font-family: cursive; font-size: 14px;">
      <li><strong>ADF Statistic</strong> – negative value, the more negative → stronger evidence of stationarity</li>
      <li><strong>p-value</strong> – if below <strong>0.05</strong>, reject the null hypothesis</li>
  </ul>

  <p>
     A low p-value indicates strong evidence against the presence of a unit root and therefore supports that 
     the <strong>time series is stationary</strong>.
  </p>

  <p style="font-size: 14px; color: green; font-weight: bold;">
      ✔ Key Insights
  </p>

  <ul style="font-family: cursive; font-size: 14px;">
      <li>Time series <strong>must be stationary</strong> before applying ARIMA</li>
      <li><strong>AIC optimization</strong> helps select the best ARIMA parameters</li>
      <li>ARIMA is effective for data with <strong>trend</strong> and <strong>seasonality</strong></li>

  <h2 style="color: green; font-size: 18px;">
    ARIMA: A Time Machine for Predictions
  </h2>

  <p>
    <strong>ARIMA</strong>, standing for <strong>Autoregressive Integrated Moving Average</strong>, 
    is a statistical model that empowers you to forecast future values based on past observations 
    in a time series dataset. It is widely used for analyzing and predicting trends in various 
    fields, from <strong>finance</strong> (stock prices) to <strong>energy forecasting</strong> 
    (temperature variations).
  </p>

  <h2 style="color: green; font-size: 16px;">
    The Building Blocks of ARIMA
  </h2>

  <p>
    ARIMA decomposes a time series into three fundamental components:
  </p>

  <ol>
    <li>
      <strong>Autoregressive (AR):</strong>  
      Captures the influence of past values on the current observation.  
      For example, when predicting tomorrow’s stock price, ARIMA considers prices from previous 
      days up to order <strong>p</strong>.
    </li>

   <li>
      <strong>Integrated (I):</strong>  
      Handles non-stationarity in the data.  
      Differencing removes trends by subtracting the previous value from the current value, 
      applied up to order <strong>d</strong>.
    </li>

  <li>
      <strong>Moving Average (MA):</strong>  
      Models the influence of past forecast errors on current predictions.  
      The parameter <strong>q</strong> specifies how many past errors are considered.
    </li>
  </ol>

  <h2 style="color: green; font-size: 16px;">
    Understanding the Notation
  </h2>

  <p>
    ARIMA models are represented as <strong>ARIMA(p, d, q)</strong>, where:
  </p>

  <ul>
    <li><strong>p</strong> – number of autoregressive terms</li>
    <li><strong>d</strong> – degree of differencing required for stationarity</li>
    <li><strong>q</strong> – number of moving average terms</li>
  </ul>

  <h1 style="color: green; font-size: 17px;">
    Example: Predicting Sales Figures
  </h1>

  <p>
    Suppose a business analyst wants to forecast monthly energy pricce using historical data.
    An ARIMA model can capture both trends and short-term dependencies.
  </p>

  <ul>
    <li>
      <strong>ARIMA(2, 1, 1):</strong>  
      Uses the past two months’ sales (<strong>p = 2</strong>),  
      applies one differencing step to remove trends (<strong>d = 1</strong>),  
      and incorporates the previous forecast error (<strong>q = 1</strong>).
    </li>
  </ul>

  <h2 style="color: green; font-size: 18px;">
    ARIMA Model Diagnostic Interpretation
  </h2>

  <h3 style="color: green;">1. Standardized Residuals Plot (Top Left)</h3>
  <b>What it shows:</b> Residuals over time (residual = actual − predicted).<br>
  <b>Why we use it:</b> To check for constant variance (homoscedasticity), time patterns,
  structural breaks, and outliers.
  <ul>
    <li><b>✅ Good:</b> Random scatter around zero with no visible pattern.</li>
    <li><b>❌ Problem:</b> Trends, cycles, or changing spread (funnel shape).</li>
  </ul>
  <b>Your plot:</b> Displays heteroskedasticity — periods of low volatility followed by
  high volatility, indicating time-varying variance.

  <h3 style="color: green;">2. Histogram + KDE Plot (Top Middle)</h3>
  <b>What it shows:</b> Distribution of residuals.<br>
  <b>Why we use it:</b> To assess the normality assumption.
  <ul>
    <li><b>✅ Good:</b> Bell-shaped curve closely matching the normal distribution.</li>
    <li><b>❌ Problem:</b> Skewness, multimodality, or heavy tails.</li>
  </ul>
  <b>Your plot:</b> Largely normal but with slightly heavy tails, suggesting occasional
  extreme deviations.

  <h3 style="color: green;">3. Q–Q Plot (Top Right)</h3>
  <b>What it shows:</b> Empirical residual quantiles vs. theoretical normal quantiles.<br>
  <b>Why we use it:</b> A more sensitive normality check than the histogram.
  <ul>
    <li><b>✅ Good:</b> Points lie on the 45° reference line.</li>
    <li><b>❌ Problem:</b> Curvature or divergence at the ends.</li>
  </ul>
  <b>Your plot:</b> Deviations at both tails indicate heavy-tailed residual behavior.

  <h3 style="color: green;">4. ACF Plot (Bottom Left)</h3>
  <b>What it shows:</b> Autocorrelation of residuals at different lags.<br>
  <b>Why we use it:</b> To test whether residuals resemble white noise.
  <ul>
    <li><b>✅ Good:</b> All bars lie within the 95% confidence bands.</li>
    <li><b>❌ Problem:</b> Significant spikes outside the bands.</li>
  </ul>
  <b>Your plot:</b> A significant spike at lag 1 indicates remaining autocorrelation.

  <h3 style="color: green;">5. PACF Plot (Bottom Middle)</h3>
  <b>What it shows:</b> Partial autocorrelation after removing intermediate lag effects.<br>
  <b>Why we use it:</b> To help identify required AR terms.
  <ul>
    <li><b>✅ Good:</b> No significant spikes.</li>
    <li><b>❌ Problem:</b> One or more dominant spikes.</li>
  </ul>
  <b>Your plot:</b> A strong spike at lag 1 suggests an AR(1) structure, despite AR(2)
  already being included.

  <h3 style="color: green;">6. Actual vs. Fitted Plot (Bottom Right)</h3>
  <b>What it shows:</b> Comparison of observed values and model predictions.<br>
  <b>Why we use it:</b> Visual inspection of model fit quality.
  <ul>
    <li><b>✅ Good:</b> Close tracking with small deviations.</li>
    <li><b>❌ Problem:</b> Systematic over- or under-prediction.</li>
  </ul>
  <b>Your plot:</b> Nearly perfect overlap, which is unusually good for real-world data

```python


def run_arima_pipeline(series,order=(2,1,1),split_date=None,freq="h"):

   
    series = series.copy()
    series.index = pd.to_datetime(series.index)
    series = series.sort_index().ffill().bfill()

    y = series

    # SPLIT
    if split_date is not None:
        train = y[y.index < split_date]
        test  = y[y.index >= split_date]
    else:
        train_size = int(len(y) * 0.8)
        train = y.iloc[:train_size]
        test  = y.iloc[train_size:]

    print("Train:", train.shape, "Test:", test.shape)

    # FIT MODEL
    model = ARIMA(train, order=order)
    model_fit = model.fit()
    print(model_fit.summary())

    results = {}

    # TEST FORECAST
    if test is not None:

        forecast_test = model_fit.forecast(steps=len(test))
        forecast_test.index = test.index

        mae = mean_absolute_error(test, forecast_test)
        rmse = np.sqrt(mean_squared_error(test, forecast_test))
        mape = np.mean(np.abs((test - forecast_test) / test)) * 100

        print("\n===== METRICS =====")
        print(f"MAE  : {mae:.4f}")
        print(f"RMSE : {rmse:.4f}")
        print(f"MAPE : {mape:.2f}%")

        plt.figure(figsize=(18,6))
        plt.plot(train.index, train, label="Train")
        plt.plot(test.index, test, label="Test")
        plt.plot(forecast_test.index, forecast_test, label="Forecast")
        plt.axvline(test.index[0], color='red', linestyle='--')
        plt.legend()
        plt.grid()
        plt.show()

        results["test_forecast"] = forecast_test

    # FUTURE FORECAST
    def forecast_future(steps, label):

        future = model_fit.forecast(steps=steps)

        future_index = pd.date_range(
            start=y.index[-1],
            periods=steps + 1,
            freq=freq)[1:]

        future.index = future_index

        plt.figure(figsize=(18,6))
        plt.plot(y.tail(24*14), label="History")
        plt.plot(future, label=f"{label} Forecast")
        plt.title(f"ARIMA - {label} Forecast")
        plt.legend()
        plt.grid()
        plt.show()

        return future

    # HORIZONS
    results["1W"] = forecast_future(24*7, "1 Week")
    results["1M"] = forecast_future(24*30, "1 Month")
    results["3M"] = forecast_future(24*90, "3 Months")

    return model_fit, results
```

            padding: 10px; background:#F2FFFF;">

  <b>ACF Plot (Autocorrelation Function)</b>
  <ul>
    <li><b>Purpose:</b> Measures correlation between a time series and its own lagged values</li>
    <li><b>Shows:</b> How much today's value correlates with past values at different time lags</li>
    <li><b>X-axis:</b> Lag number (0, 10, 20, 30, 40, 50)</li>
    <li><b>Y-axis:</b> Correlation coefficient (-1.00 to 0.75)</li>
    <li><b>What ACF Tells You:</b>
      <ul>
        <li>Decay pattern: Slow decay → non-stationarity (needs differencing)</li>
        <li>Spikes at seasonal lags → seasonal patterns</li>
        <li>MA order (q): Last significant lag helps determine MA component</li>
        <li>Trend: Persistently high correlations indicate trend</li>
      </ul>
    </li>
  </ul>

  <b>PACF Plot (Partial Autocorrelation Function)</b>
  <ul>
    <li><b>Purpose:</b> Measures correlation between a time series and its lagged values after removing effects of intermediate lags</li>
    <li><b>Shows:</b> Direct relationship between current value and specific past value</li>
    <li><b>X-axis:</b> Same lag structure</li>
    <li><b>Y-axis:</b> Partial correlation coefficient (-1.00 to 0.75)</li>
    <li><b>What PACF Tells You:</b>
      <ul>
        <li>AR order (p): Number of significant lags before cutoff</li>
        <li>Sharp cutoff → AR process</li>
        <li>Direct effects → isolates relationship at each specific lag</li>
      </ul>
    </li>
  </ul>

  <b>Key Features Shown:</b>
  <ul>
    <li>Y-axis range: -1.00 to 0.75</li>
    <li>Lags shown: Up to 50 periods</li>
    <li>Bars at each lag: Represent correlation strength</li>
    <li>Confidence bands: Horizontal dashed lines (±2/√n)</li>
  </ul>

  <b>How to Interpret for ARIMA Modeling:</b>
  <ul>
    <li><b>Pattern → Behavior → ARIMA Implication:</b>
      <ul>
        <li>AR(p) process → ACF decays gradually, PACF cuts off after lag p → Use AR(p) component</li>
        <li>MA(q) process → ACF cuts off after lag q, PACF decays gradually → Use MA(q) component</li>
        <li>ARMA(p,q) → Both decay gradually → Mixed AR and MA</li>
        <li>Non-stationary → Slow decay, stays high → Needs differencing (I)</li>
        <li>Seasonal → Spikes at seasonal lags → Seasonal ARIMA needed</li>
      </ul>
    </li>
  </ul>

  <b>For Your Energy Consumption Data:</b>
  <ul>
    <li>Check stationarity:
      <ul>
        <li>ACF decays slowly → d > 0</li>
        <li>ACF drops quickly → d = 0 (series stationary)</li>
      </ul>
    </li>
    <li>Identify AR order (p):
      <ul>
        <li>Look at PACF: Count significant lags before cutoff</li>
        <li>Example: PACF significant at lags 1,2,3 → p = 3</li>
      </ul>
    </li>
    <li>Identify MA order (q):
      <ul>
        <li>Look at ACF: Count significant lags before cutoff</li>
        <li>Example: ACF significant at lags 1,2 → q = 2</li>
      </ul>
    </li>
  </ul>

  <b>Practical Example:</b>
  <ul>
    <li>ACF: Gradually decaying over many lags</li>
    <li>PACF: Significant spike only at lag 1</li>
    <li>→ Suggests AR(1) process (p=1, d=0 or 1, q=0)</li>
  </ul>

  <p style="text-align: center; font-size: 16px; color: #00962dff; font-weight: bold;">
    ## SARIMA for Time Series Forecasting
  </p>

  <p>
    <strong>In the realm of time series analysis, predicting future values from past observations is a constant pursuit.</strong>
    SARIMA (<strong>Seasonal Autoregressive Integrated Moving Average</strong>) emerges as a powerful tool for this task,
    particularly for data exhibiting <strong>seasonality</strong>.
  </p>

  <p style="color:#005B96; font-weight:bold;">
    ## What is SARIMA?
  </p>

  <p>
    <strong>SARIMA</strong> is a statistical model that extends <strong>ARIMA</strong> by incorporating
    <strong>seasonal components</strong>. It leverages:
  </p>

  <ul>
    <li>Past observations (AR)</li>
    <li>Differencing to achieve stationarity (I)</li>
    <li>Past forecast errors (MA)</li>
    <li>Seasonal AR, seasonal differencing, and seasonal MA</li>
  </ul>

  <p style="color:#005B96; font-weight:bold;">
    ## Key Components of SARIMA
  </p>

  <ul>
    <li><strong>Non-seasonal AR (p):</strong> Influence of recent past values</li>
    <li><strong>Differencing (d):</strong> Removes long-term trends</li>
    <li><strong>Non-seasonal MA (q):</strong> Captures short-term forecast errors</li>
    <li><strong>Seasonal AR (P):</strong> Influence of past seasonal values</li>
    <li><strong>Seasonal Differencing (D):</strong> Removes seasonal trends</li>
    <li><strong>Seasonal MA (Q):</strong> Seasonal forecast error correction</li>
    <li><strong>Seasonality (s):</strong> Number of periods in one season (e.g., 24 for hourly daily seasonality)</li>
  </ul>

  <p style="color:#005B96; font-weight:bold;">
    # Example: Predicting Electricity Prices
  </p>

  <p>
    Electricity prices often exhibit <strong>strong seasonal behavior</strong>
    due to heating, cooling, and industrial demand.
    SARIMA is particularly suitable for such patterns.
  </p>

  <p>
    <strong>Example configuration:</strong>
  </p>

  <ul>
    <li>p = 2 → influence of last 2 price points</li>
    <li>d = 1 → remove long-term trend</li>
    <li>q = 1 → account for previous forecast error</li>
    <li>P = 1 → seasonal autoregression</li>
    <li>D = 1 → seasonal differencing</li>
    <li>s = 12 → monthly seasonality</li>
  </ul>

  <p>
    By fitting a SARIMA model with these parameters, future electricity prices
    can be forecasted more accurately than standard ARIMA.
  </p>

  <p style="color:#005B96; font-weight:bold;">
    ## Benefits of SARIMA
  </p>

  <ul>
    <li><strong>Excellent for seasonal data</strong></li>
    <li><strong>Interpretable statistical structure</strong></li>
    <li><strong>Strong diagnostic capabilities</strong></li>
  </ul>

  <p style="color:#005B96; font-weight:bold;">
    ## Limitations of SARIMA
  </p>

  <ul>
    <li>Requires stationarity</li>
    <li>Hyperparameter tuning can be complex</li>
    <li>Limited performance for highly non-linear patterns</li>
  </ul>

  <p>
    <strong>Conclusion:</strong><br>
    SARIMA remains a <strong>robust and interpretable model</strong> for seasonal time series forecasting,
    particularly in energy markets where demand and price patterns repeat over time.
  </p>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error

def run_sarima_pipeline(series, order=(2,1,1), seasonal_order=(1,1,1,24), split_date=None, freq="h"):

    series = series.copy()
    series.index = pd.to_datetime(series.index)
    series = series.sort_index().ffill().bfill()

    y = series

    if split_date is not None:
        train = y[y.index < split_date]
        test = y[y.index >= split_date]
    else:
        train_size = int(len(y) * 0.8)
        train = y.iloc[:train_size]
        test = y.iloc[train_size:]

    print("Train:", train.shape)
    print("Test :", test.shape)

    model = SARIMAX(
        train,
        order=order,
        seasonal_order=seasonal_order,
        enforce_stationarity=False,
        enforce_invertibility=False)

    model_fit = model.fit(disp=False)
    print(model_fit.summary())
    model_fit.plot_diagnostics(figsize=(18,10))
    plt.show()



    results = {}

    forecast_test = model_fit.forecast(steps=len(test))

    forecast_test = pd.Series(
        forecast_test.values,
        index=test.index)



    mae = mean_absolute_error(test, forecast_test)
    rmse = np.sqrt(  mean_squared_error(test, forecast_test))
    mape = np.mean(np.abs((test - forecast_test) / test)) * 100

    print("\n===== METRICS =====")
    print(f"MAE  : {mae:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAPE : {mape:.2f}%")


    plt.figure(figsize=(18,6))
    plt.plot(train.index, train, label="Train")
    plt.plot(test.index, test, label="Test")
    plt.plot(forecast_test.index, forecast_test, label="Forecast")


    plt.axvline(test.index[0],color='red',linestyle='--')
    plt.title("SARIMA Test Forecast")
    plt.legend()
    plt.grid()
    plt.show()
    results["test_forecast"] = forecast_test



    def forecast_future(steps, label):

        future = model_fit.forecast(steps=steps)
        future_index = pd.date_range(
            start=y.index[-1],
            periods=steps + 1,
            freq=freq)[1:]

        future = pd.Series( future.values,index=future_index)

        plt.figure(figsize=(18,6))
        plt.plot(y.tail(24*14),label="History")
        plt.plot( future, label=f"{label} Forecast" )

        plt.title( f"SARIMA - {label} Forecast")
        plt.legend()
        plt.grid()
        plt.show()
        return future


    results["1W"] = forecast_future(24*7, "1 Week")
    results["1M"] = forecast_future(24*30, "1 Month")
    results["3M"] = forecast_future(24*90, "3 Months")

    return model_fit, results
```

```python

def run_prophet_pipeline(series,split_date=None,freq="h"):

    # =====================
    # PREPARE DATA
    # =====================

    series = series.copy()
    series.index = pd.to_datetime(series.index)
    series = series.sort_index().ffill().bfill()

    prophet_df = pd.DataFrame({
        "ds": series.index,
        "y": series.values})

    # =====================
    # TRAIN TEST SPLIT
    # =====================

    if split_date is not None:
        train = prophet_df[prophet_df["ds"] < split_date]
        test = prophet_df[prophet_df["ds"] >= split_date]

    else:

        split_idx = int(len(prophet_df) * 0.8)
        train = prophet_df.iloc[:split_idx]
        test = prophet_df.iloc[split_idx:]

    print("Train:", train.shape)
    print("Test :", test.shape)

    # =====================
    # FIT MODEL
    # =====================

    model = Prophet(yearly_seasonality=True,weekly_seasonality=True,daily_seasonality=True,interval_width=0.95 )
    model.fit(train)

    # =====================
    # TEST FORECAST
    # =====================
    future_test = test[["ds"]]
    forecast_test = model.predict(future_test)

    y_true = test["y"]
    y_pred = forecast_test["yhat"]

    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    mape = np.mean(np.abs((y_true - y_pred) / y_true) * 100)

    print("\n===== PROPHET METRICS =====")
    print(f"MAE  : {mae:.4f}")
    print(f"RMSE : {rmse:.4f}")
    print(f"R2   : {r2:.4f}")
    print(f"MAPE : {mape:.2f}%")

    # =====================
    # PLOT TEST FORECAST
    # =====================

    plt.figure(figsize=(18,6))

    plt.plot(
        train["ds"],
        train["y"],
        label="Train")

    plt.plot(
        test["ds"],
        test["y"],
        label="Test")

    plt.plot(
        forecast_test["ds"],
        forecast_test["yhat"],
        label="Forecast")

    plt.fill_between(
        forecast_test["ds"],
        forecast_test["yhat_lower"],
        forecast_test["yhat_upper"],
        alpha=0.2,
        label="95% CI")

    plt.legend()
    plt.grid(alpha=0.3)
    plt.title("Prophet Forecast")
    plt.show()

    # =====================
    # FUTURE FORECAST
    # =====================

    future = model.make_future_dataframe( periods=24*30, freq=freq)

    forecast_future = model.predict(future)
    model.plot(forecast_future)

    plt.show()
    model.plot_components(forecast_future)
    plt.show()

    return model, forecast_future
```

```python
from statsmodels.tsa.stattools import adfuller

def check_stationarity(series):

    result = adfuller(series.dropna())

    return {
        "ADF": result[0],
        "p_value": result[1],
        "stationary": result[1] < 0.05
    }
```

```python
def run_arima(df, target="Electricity_Price_CAD_MWh"):

    df = df.tail(500)
    series = df[target].dropna()

    split = int(len(series) * 0.8)
    train, test = series[:split], series[split:]

    model = ARIMA(train, order=(1,0,1))
    fit = model.fit()
    forecast = fit.forecast(len(test))

    rmse = np.sqrt(mean_squared_error(test, forecast))

    return forecast, rmse
```

```python
def run_sarima(series):

    series = series.dropna().tail(500)

    split = int(len(series) * 0.8)
    train, test = series[:split], series[split:]


    model = SARIMAX( train,order=(1,0,1),
        seasonal_order=(0,0,1,12),
        enforce_stationarity=False,
        enforce_invertibility=False)

    fit = model.fit(disp=False)
    forecast = fit.forecast(len(test))

    rmse = np.sqrt(mean_squared_error(test, forecast))

    return forecast, rmse
```

```python
def run_seasonal(series):

    series = series.dropna().tail(500)
    horizon = int(len(series) * 0.2)
    train = series[:-horizon]

    pattern = train.groupby(train.index.hour).mean()
    forecast = [pattern[i % 24] for i in range(horizon)]
    actual = series[-horizon:].values

    rmse = np.sqrt(mean_squared_error(actual, forecast))

    return forecast, actual, rmse
```

```python
def create_sequences(data, seq=12):

    X, y = [], []
    for i in range(len(data) - seq):
        X.append(data[i:i+seq])
        y.append(data[i+seq])

    return np.array(X), np.array(y)


def run_lstm(df, col="Electricity_Price_CAD_MWh"):

    from tensorflow.keras.models import Sequential
    from tensorflow.keras.layers import LSTM, Dense

    df = df.tail(500)

    scaler = MinMaxScaler()
    data = scaler.fit_transform(df[[col]])
    X, y = create_sequences(data, seq=12)

    split = int(len(X) * 0.8)
    X_train, X_test = X[:split], X[split:]
    y_train, y_test = y[:split], y[split:]

    X_train = X_train.reshape((X_train.shape[0], 12, 1))
    X_test = X_test.reshape((X_test.shape[0], 12, 1))

    model = Sequential([ LSTM(4, input_shape=(12,1)), Dense(1)])



    model.compile(optimizer="adam", loss="mse")

    model.fit(X_train, y_train, epochs=5, batch_size=32, verbose=0)
    pred = model.predict(X_test, verbose=0)

    pred_inv = scaler.inverse_transform(pred)
    y_inv = scaler.inverse_transform(y_test.reshape(-1,1))
    rmse = np.sqrt(mean_squared_error(y_inv, pred_inv))

    return pred_inv.flatten(), rmse
```

```python
def run_full_pipeline(df):

    df = df.copy()
    df = df.sort_index().ffill().bfill()
    df = df.tail(500)

    series = df["Electricity_Price_CAD_MWh"]


    stationarity = check_stationarity(series)
    arima_forecast, arima_rmse = run_arima(df)
    sarima_forecast, sarima_rmse = run_sarima(series)
    seasonal_forecast, seasonal_actual, seasonal_rmse = run_seasonal(series)
    lstm_forecast, lstm_rmse = run_lstm(df)



    results = pd.DataFrame({

        "Model": ["ARIMA", "SARIMA", "Seasonal", "LSTM"],
        "RMSE": [arima_rmse, sarima_rmse, seasonal_rmse, lstm_rmse]}).sort_values("RMSE")


    print("\n===== RESULTS =====")
    print(results)


    # Plot
    plt.figure(figsize=(12,5))
    plt.plot(series.values[-200:], label="Actual")
    plt.plot(np.array(arima_forecast), label="ARIMA")
    plt.plot(np.array(sarima_forecast), label="SARIMA")
    plt.plot(np.array(seasonal_forecast), label="Seasonal")

    if lstm_forecast is not None:

        plt.plot(lstm_forecast, label="LSTM")
    plt.legend()
    plt.title("Model Comparison (500 rows)")
    plt.grid()
    plt.show()

    return results, stationarity
```

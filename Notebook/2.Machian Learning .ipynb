```python
def prepare_data(df1,target="Electricity_Price_CAD_MWh"):

    df1=df1.sort_index()

    # missing
    df1=df1.ffill().bfill()

    y=df1[target]

    X=df1.select_dtypes(include="number").copy()

    leakage_cols=[
        "Energy_Demand_MW",
        "Energy_Price_CAD",
        "Electricity_Price_CAD_MWh",
        "Gas_Price_CAD_GJ"]

    X=X.drop(columns=leakage_cols,errors="ignore")

    # =========================
    # IMPORTANT FIX (NaN + Inf)
    # =========================

    X=X.replace([np.inf,-np.inf],np.nan)
    X=X.fillna(0)
    y=y.ffill().bfill()

    # split
    split=int(len(df1)*0.8)

    X_train=X.iloc[:split]
    X_test=X.iloc[split:]

    y_train=y.iloc[:split]
    y_test=y.iloc[split:]

    return X_train,X_test,y_train,y_test
```

```python
def evaluate(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    return mae, rmse, r2
```

```python
def run_rf(X_train, X_test, y_train, y_test):

    rf = RandomForestRegressor(
        n_estimators=200,
        max_depth=12,
        random_state=42)

    rf.fit(X_train, y_train)
    pred = rf.predict(X_test)
    mae,rmse,r2=evaluate(y_test,pred)

    print("RF MAE:",mae)
    print("RF RMSE:",rmse)
    print("RF R2:",r2)

    return rf
```

```python

def prepare_data(df, HORIZON=168, TARGET="Electricity_Price_CAD_MWh" ):

    df=df.copy()
    df=df.sort_index()
    df=df.ffill().bfill()

    # future target (7 days ahead)
    y=df[TARGET].shift(-HORIZON)

    # features
    X=df.select_dtypes(include=np.number).copy()

    leakage_cols=[
        "Energy_Demand_MW",
        "Energy_Price_CAD",
        "Electricity_Price_CAD_MWh",
        "Gas_Price_CAD_GJ"]

    X=X.drop(columns=leakage_cols,errors="ignore")

    if TARGET in X.columns:
        X=X.drop(columns=[TARGET])

    # align safely
    data=X.copy()
    data["target"]=y

    data=data.dropna(subset=["target"])

    X=data.drop(columns=["target"])
    y=data["target"]

    # fix NaN/Inf
    X=X.replace([np.inf,-np.inf],np.nan)
    X=X.fillna(0)

    split=int(len(X)*0.8)

    X_train=X.iloc[:split]
    X_test=X.iloc[split:]

    y_train=y.iloc[:split]
    y_test=y.iloc[split:]

    return X_train,X_test,y_train,y_test,X,data
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python

def train_model(X_train,y_train,X_test,y_test):

    model=xgb.XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42)

    model.fit(X_train,y_train)
    preds=model.predict(X_test)


    mae=mean_absolute_error(y_test,preds)
    rmse=np.sqrt(mean_squared_error(y_test,preds))
    r2=r2_score(y_test,preds)

    print("\n===== TEST RESULTS =====")
    print("MAE :",mae)
    print("RMSE:",rmse)
    print("R2  :",r2)

    return model,preds
```

```python
# =========================
# FORECAST FUTURE 168h
# =========================

def forecast_next_7days(model,X,df, HORIZON=168,):

    future_X=X.tail(HORIZON)
    future_pred=model.predict(future_X)

    future_dates=pd.date_range(
        start=df.index[-1],
        periods=HORIZON,
        freq="h")

    future_df=pd.DataFrame(
        {"Forecast":future_pred},
        index=future_dates)

    return future_df
```

```python
# =========================
# PLOT FUNCTION
# =========================

def plot_results(y_test,preds,future_df):

    plt.figure(figsize=(16,6))
    plt.plot(y_test.values,label="Actual")
    plt.plot(preds,label="Predicted",linestyle="--")

    plt.title("Test Prediction (XGBoost)")
    plt.legend()
    plt.grid()
    plt.show()


    plt.figure(figsize=(16,6))
    plt.plot(future_df.index,future_df["Forecast"])
    plt.title("Next 7 Days Forecast (168h)")
    plt.grid()
    plt.show()
```

```python
X_train,X_test,y_train,y_test,X,df=prepare_data(df1)
model,preds=train_model(X_train,y_train,X_test,y_test)
future_df=forecast_next_7days(model,X,df)

plot_results(y_test,preds,future_df)

print("\n===== FUTURE FORECAST =====")
print(future_df.head())
```

```python

target = "Electricity_Price_CAD_MWh"

y = df[target]
X = df.drop(columns=[
    "Electricity_Price_CAD_MWh",
    "Energy_Demand_MW",
    "Gas_Price_CAD_GJ"], errors="ignore")



train_size = int(len(df) * 0.8)

X_train = X.iloc[:train_size]
X_test  = X.iloc[train_size:]

y_train = y.iloc[:train_size]
y_test  = y.iloc[train_size:]

print("Train:", X_train.shape, "Test:", X_test.shape)
```

```python
# =========================================================
# MODELS
# =========================================================

models = {

    "RandomForest": RandomForestRegressor(
        n_estimators=200,
        max_depth=12,
        random_state=42),

    "XGBoost": xgb.XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42),

    "GradientBoosting": GradientBoostingRegressor(
        n_estimators=200,
        learning_rate=0.05,
        max_depth=3,
        random_state=42)}
```

```python

# =========================================================
# TRAIN + PREDICT
# =========================================================

predictions = {}
results = []

for name, model in models.items():
    print(f"\nTraining {name}...")

    # train
    model.fit(X_train, y_train)
    # predict
    preds = model.predict(X_test)
    # save predictions
    predictions[name] = preds

    # metrics
    mae = mean_absolute_error(y_test, preds)
    rmse = np.sqrt(mean_squared_error(y_test, preds))
    r2 = r2_score(y_test, preds)

    # save results
    results.append({
        "Model": name,
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "R2": round(r2, 3)})
```

```python

# =========================================================
# RESULTS TABLE
# =========================================================

results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by="RMSE")

print("\n===== MODEL COMPARISON =====\n")
print(results_df)
```

```python

# =========================================================
# PLOT: ACTUAL VS PREDICTIONS
# =========================================================

plt.figure(figsize=(16,6))

# actual values
plt.plot(
    y_test.values,
    label="Actual",
    linewidth=3)

# model predictions
for name, preds in predictions.items():

    plt.plot(
        preds,
        label=name,
        alpha=0.8)

plt.title("Electricity Price Forecast Comparison")
plt.xlabel("Time")
plt.ylabel("Price")
plt.legend()
plt.grid(True)
plt.show()
```

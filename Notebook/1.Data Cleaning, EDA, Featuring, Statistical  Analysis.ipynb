```python
# Calculate IQR
Q1 = df["Electricity_Price_CAD_MWh"].quantile(0.25)
Q3 = df["Electricity_Price_CAD_MWh"].quantile(0.75)
IQR = Q3 - Q1

# Filter in place (keep only non-outliers)
df = df[(df["Electricity_Price_CAD_MWh"] >= (Q1 - 1.5 * IQR)) & 
        (df["Electricity_Price_CAD_MWh"] <= (Q3 + 1.5 * IQR))]

# Reset index (optional)
df = df.reset_index(drop=True)

print(f"Dataset after outlier removal: {df.shape}")
```

```python
def remove_outliers_iqr(df, column, multiplier=3):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - multiplier * IQR
    upper_bound = Q3 + multiplier * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    df_clean = df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]
    print(f"  {column}: {len(outliers)} outlier removed ({(len(outliers)/len(df))*100:.2f}%)")
    return df_clean

# حذف پرت‌ها از ستون‌های کلیدی
df = remove_outliers_iqr(df, 'Total', multiplier=3)
df = remove_outliers_iqr(df, 'Unit price', multiplier=5)
df = remove_outliers_iqr(df, 'Quantity', multiplier=3)
```

```python
def statistical_summary(df):
  
    target = "Electricity_Price_CAD_MWh"
    
    price = df[target]
    demand = df["AIL_Total_Demand_MW"]
    
    # Core statistics
    stats_dict = {
        "mean_demand": demand.mean(),
        "max_demand": demand.max(),
        "min_demand": demand.min(),
        "std_demand": demand.std(),
        "range_demand" :demand.max()-demand.min(),
        
        "mean_price": price.mean(),
        "std_price": price.std(),
        "min_price": price.min(),
        "max_price": price.max(),
        "range_price": price.max() - price.min(),
        
        "median_price": price.median(),
        "q1_price": price.quantile(0.25),
        "q3_price": price.quantile(0.75),
        
        "skewness": price.skew(),
        "kurtosis": price.kurtosis(),
        
        "volatility": price.std(),
        "cv": price.std() / price.mean() if price.mean() > 0 else 0}
    

    # Value at Risk (VaR)
    stats_dict["var_95"] = stats_dict["mean_price"] - (1.65 * stats_dict["std_price"])
    stats_dict["var_99"] = stats_dict["mean_price"] - (2.33 * stats_dict["std_price"])
    
    df_stats = pd.DataFrame([stats_dict])
    
    # Distribution insight
    skew = stats_dict["skewness"]
    
    if pd.isna(skew):  # if NaN
        distribution = "No distribution information available"
    elif skew > 0.5:
        distribution = "Right-skewed (high price spikes dominate market)"
    elif skew < -0.5:
        distribution = "Left-skewed (low price pressure dominates market)"
    else:
        distribution = "Approximately symmetric market behavior"
    

    df_stats .head()
    
    return df_stats
```

```python
def time_series_analysis(df):

    df = df.copy()
    df = df.sort_index()
    df = df[~df.index.duplicated()]

    # =========================
    # Hourly Price Pattern
    # =========================
    hourly_avg = df.groupby(df.index.hour)['Electricity_Price_CAD_MWh'].mean()

    plt.figure(figsize=(10, 6))
    sns.lineplot(x=hourly_avg.index, y=hourly_avg.values)
    plt.title("Average Hourly Price Pattern")
    plt.xlabel("Hour of Day")
    plt.ylabel("Average Price ($/MWh)")
    plt.xticks(range(0, 24))
    plt.show()

    # =========================
    # Weekly Pattern
    # =========================
    weekday_avg = df.groupby(df.index.dayofweek)['Electricity_Price_CAD_MWh'].mean()

    plt.figure(figsize=(10, 6))
    sns.lineplot(x=weekday_avg.index, y=weekday_avg.values)
    plt.title("Average Weekly Price Pattern")
    plt.xlabel("Day of Week")
    plt.ylabel("Average Price ($/MWh)")
    plt.xticks(range(0, 7))
    plt.show()

    # =========================
    # Yearly Trend
    # =========================
    yearly = df.groupby(df.index.year)["AIL_Total_Demand_MW"].sum()

    yearly.plot(kind="line", figsize=(10, 5))
    plt.title("Yearly Energy Consumption")
    plt.xlabel("Year")
    plt.ylabel("Total Consumption (MW)")
    plt.show()

    # =========================
    # Price vs Demand
    # =========================
    plt.figure(figsize=(10, 6))
    sns.scatterplot(
        x=df['AIL_Total_Demand_MW'],
        y=df['Electricity_Price_CAD_MWh'])
    plt.title("Price vs Demand")
    plt.xlabel("Hourly Demand")
    plt.ylabel("Price ($/MWh)")
    plt.show()

    # =========================
    # Monthly Trend (Consumption)
    # =========================
    monthly_avg = df.groupby(df.index.month)['AIL_Total_Demand_MW'].mean()

    plt.figure(figsize=(10,6))
    sns.barplot(
        x=monthly_avg.index,
        y=monthly_avg.values,
        palette='viridis')

    plt.title('Average Energy Consumption by Month')
    plt.xlabel('Month')
    plt.ylabel('Average Energy Consumption (MW)')

    plt.xticks(
        ticks=range(12),
        labels=['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec'])
    plt.show()

    # =========================
    # Gas Price Trend
    # =========================
    monthly_gas = df.groupby(df.index.month)['Gas_Price_CAD_GJ'].mean()

    plt.figure(figsize=(10, 6))
    plt.plot(monthly_gas.index,
             monthly_gas.values,
             marker='o',
             color='b')

    plt.title('Gas Price Trend by Month')
    plt.xlabel('Month')
    plt.ylabel('Gas Price (CAD/GJ)')
    plt.xticks(range(1, 13))
    plt.grid(alpha=0.5)
    plt.show()

    return df
```

```python

def gas_behavior_plot(df):

    # =========================
    # Data Cleaning
    # =========================
    df = df.copy()
    df = df.sort_index()
    df = df[~df.index.duplicated(keep="first")]
    df = df.dropna(subset=["Gas_Price_CAD_GJ"])
    df["Hour"] = df.index.hour
    df["DayOfWeek"] = df.index.day_name()
    df["month_name"] = df.index.month_name()

    # ====================================================
    # 1. Hourly Gas Price by Day of Week
    # ====================================================
    plt.figure(figsize=(12, 6))

    sns.lineplot(
        data=df,
        x="Hour",
        y="Gas_Price_CAD_GJ",
        hue="DayOfWeek",
        palette="viridis",
        errorbar=None)

    plt.title("Gas Price by Hour and Day of Week")
    plt.xlabel("Hour of Day")
    plt.ylabel("Gas Price (CAD/GJ)")
    plt.legend(title="Day of Week", bbox_to_anchor=(1.05, 1))
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ====================================================
    # 2. Monthly Trend by Year
    # ====================================================
    monthly_sales = (df.groupby(['Year', 'Month'])['Gas_Price_CAD_GJ'].mean().reset_index())

    plt.figure(figsize=(12, 6))
    years = monthly_sales['Year'].unique()
    colors = itertools.cycle([ 'red', 'green', 'blue', 'orange', 'purple', 'brown'])

    for year in years:
        year_data = monthly_sales[monthly_sales['Year'] == year]

        plt.plot(
            year_data['Month'],
            year_data['Gas_Price_CAD_GJ'],
            marker='o',
            color=next(colors),
            label=str(year) )

    plt.xlabel('Month')
    plt.ylabel('Average Gas Price (CAD/GJ)')
    plt.title('Monthly Gas Price Trend')

    month_names = [
        'Jan', 'Feb', 'Mar', 'Apr',
        'May', 'Jun', 'Jul', 'Aug',
        'Sep', 'Oct', 'Nov', 'Dec']

    plt.xticks(range(1, 13), month_names)
    plt.legend(title="Year")
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

    # ====================================================
    # 3. Average Monthly Gas Price
    # ====================================================
    month_order = [
        'January', 'February', 'March',
        'April', 'May', 'June','July', 'August', 'September', 'October', 'November', 'December']

    monthly_avg = (df.groupby('month_name')['Gas_Price_CAD_GJ'].mean().reindex(month_order))

    bright_palette = sns.color_palette("bright", 12)
    plt.figure(figsize=(14, 7))

    bars = plt.bar(
        monthly_avg.index,
        monthly_avg.values,
        color=bright_palette)

    plt.title('Average Monthly Gas Price (CAD/GJ)', fontsize=16,fontweight='bold')

    plt.xlabel('Month')
    plt.ylabel('Gas Price (CAD/GJ)')
    plt.xticks(rotation=45)

    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height,
            f'{height:.2f}',
            ha='center',
            va='bottom',
            fontweight='bold')

    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()

    return df
```

```python
def seasonal_market_analysis(df):

    df = df.copy()

    # =========================
    # Monthly Price Pattern
    # =========================
    season_avg_consumption = df.groupby('Season')['Electricity_Price_CAD_MWh'].mean().reset_index()
    # Line chart
    plt.figure(figsize=(10, 6))
    plt.plot(season_avg_consumption['Season'], season_avg_consumption['Electricity_Price_CAD_MWh'], marker='o', linestyle='-', color='b')
    plt.title('Average Electricity_Price_CAD_MWh by Season', fontsize=16)
    plt.xlabel('Season', fontsize=14)
    plt.ylabel('Average Electricity_Price_CAD_MWh', fontsize=14)
    plt.grid(visible=True, linestyle='--', alpha=0.6)
    plt.show()
    # =========================
    # Temperature Impact
    # =========================
    temp_avg = None

    if "Temperature_C" in df.columns:
        temp_bins = pd.cut(df["Temperature_C"], bins=10)
        temp_avg = df.groupby(temp_bins)['Electricity_Price_CAD_MWh'].mean()

        plt.figure(figsize=(10, 6))
        plt.plot(range(len(temp_avg)), temp_avg.values, marker='o')
        plt.title("Price vs Temperature")
        plt.xlabel("Temperature Range")
        plt.ylabel("Average Price ($/MWh)")
        plt.xticks(
            range(len(temp_avg)),
            [str(b.left) for b in temp_avg.index],
            rotation=45)
        plt.show()

    # =========================
    # Seasonal Analysis
    # =========================
    if 'Season' in df.columns:

        if "Total_Cost" not in df.columns:
            df["Total_Cost"] = (df["AIL_Total_Demand_MW"] * df["Electricity_Price_CAD_MWh"])

        season_mean = df.groupby("Season")["AIL_Total_Demand_MW"].mean()
        season_max = df.groupby("Season")["AIL_Total_Demand_MW"].max()
        season_cost_mean = df.groupby("Season")["Total_Cost"].mean()
        season_cost_sum = df.groupby("Season")["Total_Cost"].sum()

        season_mean.plot(kind="bar", figsize=(8,5), title="Seasonal Energy Consumption")
        plt.show()

        season_max.plot(kind="bar", figsize=(8,5), title="Seasonal Peak Load")
        plt.show()

        season_cost_mean.plot(kind="bar", figsize=(8,5), title="Seasonal Cost (Mean)")
        plt.show()

        season_cost_sum.plot(kind="bar", figsize=(8,5), title="Seasonal Total Energy Cost")
        plt.show()


    energy_counts = df['Season'].value_counts()
    funding_counts = df['Peak_Hour'].value_counts() 

    fig, ax = plt.subplots(1, 2, figsize=(14, 7)) 

    ax[0].pie(energy_counts, labels=energy_counts.index, autopct='%1.1f%%', startangle=140)
    ax[0].set_title('Seasonal') 
    ax[1].pie(funding_counts, labels=funding_counts.index, autopct='%1.1f%%', startangle=140)
    ax[1].set_title('Peak Hour') 
    plt.tight_layout()
    plt.show()

    return df
```

```python
def behavioral(df):


    df = df.copy()

    # =========================
    # CLEAN DATA (IMPORTANT FIX)
    # =========================
    df = df.sort_index()
    df = df[~df.index.duplicated(keep="first")]

    # drop rows where key columns are missing (fix scatterplot crash)
    df = df.dropna(subset=[
        "AIL_Total_Demand_MW",
        "Electricity_Price_CAD_MWh",
        "Gas_Price_CAD_GJ"])

    # =========================
    # Hourly Demand Pattern
    # =========================
    hour_pattern = df.groupby(df.index.hour)["AIL_Total_Demand_MW"].mean()

    plt.figure(figsize=(10, 5))
    hour_pattern.plot(kind="line", marker="o")
    plt.title("Hourly Demand Pattern")
    plt.xlabel("Hour of Day")
    plt.ylabel("Average Demand (MW)")
    plt.xticks(range(0, 24))
    plt.grid(True)
    plt.show()

    # =========================
    # Weekday Pattern
    # =========================
    df["Weekday_Name"] = df.index.day_name()

    weekday_order = [
        "Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday"]

    weekday_pattern = df.groupby("Weekday_Name")["AIL_Total_Demand_MW"].mean().reindex(weekday_order)

    plt.figure(figsize=(10, 5))
    weekday_pattern.plot(kind="bar", color="skyblue")
    plt.title("Weekday Consumption Pattern")
    plt.xlabel("Day")
    plt.ylabel("Average Demand (MW)")
    plt.xticks(rotation=45)
    plt.grid(axis="y", alpha=0.3)
    plt.show()

    # =========================
    # Cheapest Hours (Price)
    # =========================
    cheap_hours = df.groupby(df.index.hour)["Electricity_Price_CAD_MWh"].min()

    plt.figure(figsize=(10, 5))
    cheap_hours.plot(kind="bar", color="green")
    plt.title("Average Electricity Price by Hour")
    plt.xlabel("Hour")
    plt.ylabel("Price (CAD/MWh)")
    plt.xticks(rotation=0)
    plt.grid(axis="y", alpha=0.3)
    plt.show()

    # =========================
    # Peak Cost Hours
    # =========================
    df["Total_Cost_Hour"] = (df["AIL_Total_Demand_MW"] * df["Electricity_Price_CAD_MWh"])

    peak_hours = df.groupby(df.index.hour)["Total_Cost_Hour"].mean()

    plt.figure(figsize=(10, 5))
    peak_hours.plot(kind="bar", color="red")
    plt.title("Average Cost by Hour")
    plt.xlabel("Hour")
    plt.ylabel("Cost (MW × Price)")
    plt.grid(axis="y", alpha=0.3)
    plt.show()

    # =========================
    # Electricity Price Trend
    # =========================
    price_trend = df.groupby(df.index.year)["Electricity_Price_CAD_MWh"].mean()

    plt.figure(figsize=(8, 4))
    price_trend.plot(marker="o")
    plt.title("Electricity Price Trend (Yearly)")
    plt.xlabel("Year")
    plt.ylabel("Avg Price (CAD/MWh)")
    plt.grid(True)
    plt.show()

    # =========================
    # Gas Price Trend
    # =========================
    gas_trend = df.groupby(df.index.year)["Gas_Price_CAD_GJ"].mean()

    plt.figure(figsize=(8, 4))
    gas_trend.plot(marker="o", color="orange")
    plt.title("Gas Price Trend (Yearly)")
    plt.xlabel("Year")
    plt.ylabel("Avg Gas Price (CAD/GJ)")
    plt.grid(True)
    plt.show()

    # =========================
    # Gas vs Electricity (SAFE SCATTER)
    # =========================
    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        x=df["Gas_Price_CAD_GJ"],
        y=df["Electricity_Price_CAD_MWh"])
    plt.title("Gas vs Electricity Price Relationship")
    plt.xlabel("Gas Price (CAD/GJ)")
    plt.ylabel("Electricity Price (CAD/MWh)")
    plt.grid(True)
    plt.show()

    # =========================
    # Demand vs Price
    # =========================
    plt.figure(figsize=(8, 5))
    sns.scatterplot(
        x=df["Electricity_Price_CAD_MWh"],
        y=df["AIL_Total_Demand_MW"])
    plt.title("Demand vs Electricity Price")
    plt.xlabel("Price (CAD/MWh)")
    plt.ylabel("Demand (MW)")
    plt.grid(True)
    plt.show()

    return df
```

```python

def regression_analysis(df, target_column):

    results = []

    numeric_df = df.select_dtypes(include=np.number)

    feature_columns = [
        col for col in numeric_df.columns
        if col != target_column]

    for feature in feature_columns:

        temp_df = df[[feature, target_column]].dropna()

        if len(temp_df) < 5:
            continue

        x = temp_df[feature].astype(float)
        y = temp_df[target_column].astype(float)

        slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)

        results.append({
            "Feature": feature,
            "Slope": slope,
            "R_squared": r_value**2,
            "P_value": p_value,
            "Std_error": std_err})

    results_df = pd.DataFrame(results)

    results_df = results_df.sort_values(
        by="R_squared",
        ascending=False)

    return results_df
```

```python


def run_anova(df, group_col, target_col):

    df_clean = df[[group_col, target_col]].dropna()

    groups = []

    for g in df_clean[group_col].unique():
        groups.append(
            df_clean[df_clean[group_col] == g][target_col].values)

    # ANOVA test
    f_stat, p_value = f_oneway(*groups)

    print(f"Group: {group_col} -> Target: {target_col}")
    print(f"F-statistic: {f_stat:.4f}")
    print(f"P-value: {p_value:.6f}")

    return {
        "f_stat": f_stat,
        "p_value": p_value,
        "significant": p_value < 0.05}
```

```python


def run_chi_square(df, col1, col2):

    df_clean = df[[col1, col2]].dropna()
    df_clean = df_clean.drop_duplicates()

    # contingency table
    table = pd.crosstab(df_clean[col1], df_clean[col2])

    chi2, p, dof, expected = chi2_contingency(table)

    print("Chi-square test")
    print(f"Variables: {col1} vs {col2}")
    print(f"Chi2: {chi2:.4f}")
    print(f"P-value: {p:.6f}")
    print(f"DOF: {dof}")

    return {
        "chi2": chi2,
        "p_value": p,
        "dof": dof,
        "significant": p < 0.05}
```

```python

def  run_statistical_analysis(df, target):

    df = df.copy()

    # -------------------------
    # CLEAN DATA
    # -------------------------
    df = df[~df.index.duplicated(keep="first")]
    df = df.reset_index(drop=True)

    # detect columns
    categorical_cols = df.select_dtypes(include=["object", "category"]).columns
    numeric_cols = df.select_dtypes(include=np.number).columns

    # remove target from features
    numeric_features = [c for c in numeric_cols if c != target]

    # =========================
    # 1. ANOVA
    # =========================
    anova_results = []

    for col in categorical_cols:

        groups = [
            df[df[col] == g][target].dropna().values
            for g in df[col].dropna().unique()]

        if len(groups) > 1:
            try:
                f_stat, p_val = f_oneway(*groups)

                anova_results.append([
                    col, f_stat, p_val, p_val < 0.05])
            except:
                continue

    anova_df = pd.DataFrame(
        anova_results,
        columns=["Feature", "F_stat", "P_value", "Significant"]).sort_values("P_value")

    # =========================
    # 2. CHI-SQUARE
    # =========================
    chi_results = []

    cat_list = list(categorical_cols)

    for i in range(len(cat_list)):
        for j in range(i + 1, len(cat_list)):

            col1, col2 = cat_list[i], cat_list[j]

            table = pd.crosstab(df[col1], df[col2])

            if table.shape[0] > 1 and table.shape[1] > 1:

                try:
                    chi2, p, _, _ = chi2_contingency(table)

                    chi_results.append([
                        f"{col1} vs {col2}",
                        chi2, p, p < 0.05])

                except:
                    continue

    chi_df = pd.DataFrame(chi_results, columns=["Pair", "Chi2", "P_value", "Significant"]).sort_values("P_value")

    # =========================
    # 3. LINEAR REGRESSION
    # =========================
    reg_results = []

    for feature in numeric_features:

        temp = df[[feature, target]].dropna()

        if len(temp) < 5:
            continue

        try:
            X = temp[feature].values.reshape(-1, 1)
            y = temp[target].values

            model = LinearRegression()
            model.fit(X, y)

            y_pred = model.predict(X)

            reg_results.append([
                feature,
                model.coef_[0],
                model.intercept_,
                r2_score(y, y_pred)])

        except:
            continue

    reg_df = pd.DataFrame(reg_results,columns=["Feature", "Slope", "Intercept", "R2"]).sort_values("R2", ascending=False)

    # =========================
    # RETURN ALL RESULTS
    # =========================
    return {
        "anova": anova_df,
        "chi_square": chi_df,
        "regression": reg_df}


# =========================
# PRETTY PRINT FUNCTION
# =========================
def pretty_report(results):

    print("\n" + "="*70)
    print(" STATISTICAL ANALYSIS REPORT")
    print("="*70)

    print("\n ANOVA (categorical → target)")
    print(results["anova"])

    print("\n CHI-SQUARE (categorical ↔ categorical)")
    print(results["chi_square"])

    print("\n LINEAR REGRESSION (numeric → target)")
    print(results["regression"])

 
```

# <p style="background: linear-gradient(to right,white,green);color: transparent;font-family: Gabriola;color: navy;font-size: 80%;text-align: center;border-radius: 10px 70px"> Anomaly, Risk & Spark Spread Analysis</p>

```python
def calculate_market_indicators(df, price_col='Electricity_Price_CAD_MWh', ai_consultant=None):

    df = df.copy()
    
 
    df['hour'] = df.index.hour
    df['weekday'] = df.index.dayofweek
    df['month'] = df.index.month
    df['is_weekend'] = (df['weekday'] >= 5).astype(int)
    
    # ============================================================
    # 1.  (Anomaly Detection)
    # ============================================================
    prices = df[price_col]
    
    #  Z-Score 
    z_scores = np.abs((prices - prices.mean()) / prices.std())
    anomaly_mask_z = z_scores > 3
    
    #  IQR 
    q1, q3 = prices.quantile([0.25, 0.75])
    iqr = q3 - q1
    anomaly_mask_iqr = (prices < q1 - 1.5 * iqr) | (prices > q3 + 1.5 * iqr)
    
  
    anomaly_mask = anomaly_mask_z | anomaly_mask_iqr
    anomalies = prices[anomaly_mask]
    

    anomaly_stats = {
        "total_anomalies": int(anomaly_mask.sum()),
        "anomaly_ratio_pct": round(anomaly_mask.mean() * 100, 2),
        "mean_anomaly_price": round(anomalies.mean(), 2) if len(anomalies) > 0 else 0,
        "max_anomaly_price": round(anomalies.max(), 2) if len(anomalies) > 0 else 0,
        "min_anomaly_price": round(anomalies.min(), 2) if len(anomalies) > 0 else 0}
    
    
    if len(anomalies) > 0:
        df_anom = df.copy()
        df_anom['is_anomaly'] = anomaly_mask
        anomaly_stats["common_anomaly_hours"] = (
            df_anom[df_anom['is_anomaly']]['hour']
            .value_counts()
            .head(3)
            .index.tolist())
    else:
        anomaly_stats["common_anomaly_hours"] = []
    
    # ============================================================
    # 2. Spark Spread Analysis
    # ============================================================
    if 'Gas_Price_CAD_GJ' in df.columns:
        heat_rate = 7.0  
        df['Spark_Spread'] = df[price_col] - (df['Gas_Price_CAD_GJ'] * heat_rate)
        
        spark_stats = {
            "avg_spark_spread": round(df['Spark_Spread'].mean(), 2),
            "max_spark_spread": round(df['Spark_Spread'].max(), 2),
            "min_spark_spread": round(df['Spark_Spread'].min(), 2),
            "negative_spread_pct": round((df['Spark_Spread'] < 0).mean() * 100, 2),
            "positive_spread_pct": round((df['Spark_Spread'] > 0).mean() * 100, 2)}
        
        profitable_hours = (
            df[df['Spark_Spread'] > 0]
            .groupby('hour')
            .size()
            .nlargest(3)
            .index.tolist())
        spark_stats["profitable_hours"] = profitable_hours
        
        spark_by_hour = df.groupby('hour')['Spark_Spread'].mean()
        spark_stats["best_generation_hours"] = spark_by_hour.nlargest(3).index.tolist()
        spark_stats["worst_generation_hours"] = spark_by_hour.nsmallest(3).index.tolist()
        
    else:
        spark_stats = {
            "avg_spark_spread": 0,
            "max_spark_spread": 0,
            "min_spark_spread": 0,
            "negative_spread_pct": 0,
            "positive_spread_pct": 0,
            "profitable_hours": [],
            "best_generation_hours": [],
            "worst_generation_hours": []}
    
    # ============================================================
    # 3.   (Volatility & Risk Analysis)
    # ============================================================
   
    cv = (df[price_col].std() / df[price_col].mean()) * 100
    
  
    if cv < 15:
        risk_level = "LOW"
        risk_color = "🟢"
    elif cv < 30:
        risk_level = "MODERATE"
        risk_color = "🟡"
    else:
        risk_level = "HIGH"
        risk_color = "🔴"
    
   
    intraday_volatility = df.groupby('Hour')[price_col].std().mean()
    daily_avg = df.groupby(df.index.date)[price_col].mean()
    interday_volatility = daily_avg.std()
    volatility_ratio = intraday_volatility / (interday_volatility + 1e-6)
    
  
    rolling_volatility = df[price_col].rolling(window=24).std()
    current_volatility = rolling_volatility.iloc[-1] if len(rolling_volatility) > 0 else 0
    
    volatility_stats = {
        "cv_pct": round(cv, 2),
        "risk_level": risk_level,
        "risk_color": risk_color,
        "intraday_volatility": round(intraday_volatility, 2),
        "interday_volatility": round(interday_volatility, 2),
        "volatility_ratio": round(volatility_ratio, 2),
        "current_volatility": round(current_volatility, 2),
        "volatility_percentile": round((cv / 100) * 100, 1)}
    
   
    
    results = {
        'anomaly': anomaly_stats,
        'spark_spread': spark_stats,
        'volatility': volatility_stats}
        

    
```

```python
calculate_market_indicators(df, price_col='Electricity_Price_CAD_MWh')
```
